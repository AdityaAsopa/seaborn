```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
sns.set_theme()
```

```python
g = sns.BrokenAxes(ylims=[(0, 50), (450, 500)])
print(g.ax_bottom.get_ylim(), g.ax_top.get_ylim())
```

```python
healthexp = sns.load_dataset("healthexp")
g = sns.BrokenAxes(ylims=[(0, 7500), (9000, 12500)])
g.plot(sns.lineplot, data=healthexp, x="Year", y="Spending_USD", hue="Country")
g.set_axis_labels("Year", "Health Spending (USD)")
```

```python
g = sns.BrokenAxes(ylims=[(0, 7500), (9000, 12500)], height_ratios=[4, 1])
g.plot(sns.lineplot, data=healthexp, x="Year", y="Spending_USD", hue="Country")
g.set_axis_labels("Year", "Health Spending (USD)")
```

```python
g = sns.BrokenAxes(ylims=[(0, 7500), (9000, 12500)], height_ratios=[4, 1])
usa = healthexp[healthexp["Country"] == "USA"]
others = healthexp[healthexp["Country"] != "USA"]
for ax in g.axes:
    sns.lineplot(data=others, x="Year", y="Spending_USD",
                 hue="Country", ax=ax, legend=(ax is g.ax_bottom))
    sns.lineplot(data=usa, x="Year", y="Spending_USD",
                 color="crimson", linewidth=2.5, label="USA",
                 ax=ax, legend=(ax is g.ax_bottom))
g.set_axis_labels("Year", "Health Spending (USD)")
```

```python
latest = healthexp[healthexp["Year"] == 2020]
g = sns.BrokenAxes(ylims=[(0, 7500), (9000, 12500)], height_ratios=[3, 1])
g.plot(sns.barplot, data=latest, x="Country", y="Spending_USD")
g.set_axis_labels("Country", "Health Spending (USD, 2020)")
```

```python
g = sns.BrokenAxes(
    ylims=[(0, 2000), (3500, 7500), (9000, 12500)],
    height_ratios=[1, 3, 1],
)
g.plot(sns.lineplot, data=healthexp, x="Year", y="Spending_USD", hue="Country")
g.set_axis_labels("Year", "Health Spending (USD)")
```

```python
g = sns.BrokenAxes(xlims=[(1970, 1985), (2005, 2020)])
g.plot(sns.lineplot, data=healthexp, x="Year", y="Spending_USD", hue="Country")
g.set_axis_labels("Year", "Health Spending (USD)")
```

```python
g = sns.BrokenAxes(
    xlims=[(0, 7000), (9000, 12500)],
    width_ratios=[4, 1],
    wspace=0.08,
)
g.plot(
    sns.scatterplot,
    data=healthexp,
    x="Spending_USD",
    y="Life_Expectancy",
    hue="Country",
)
g.set_axis_labels("Health Spending (USD)", "Life Expectancy (years)")
```

```python
g = sns.BrokenAxes(
    ylims=[(60, 80), (85, 90)],
    xlims=[(0, 7000), (9000, 12500)],
    height_ratios=[3, 1],
    width_ratios=[4, 1],
)
g.plot(
    sns.scatterplot,
    data=healthexp,
    x="Spending_USD",
    y="Life_Expectancy",
    hue="Country",
)
g.set_axis_labels("Health Spending (USD)", "Life Expectancy (years)")
```

```python
fig = plt.figure(figsize=(12, 4))
gs = gridspec.GridSpec(1, 2, figure=fig, wspace=0.35)

# Left panel: broken y-axis
g = sns.BrokenAxes(
    ylims=[(0, 7500), (9000, 12500)],
    height_ratios=[4, 1],
    subplot_spec=gs[0, 0],
)
g.plot(sns.lineplot, data=healthexp, x="Year", y="Spending_USD", hue="Country")
g.set_axis_labels("Year", "Spending (USD)")

# Right panel: ordinary axis for comparison
ax = fig.add_subplot(gs[0, 1])
sns.lineplot(data=healthexp, x="Year", y="Life_Expectancy", hue="Country", ax=ax)
ax.set(xlabel="Year", ylabel="Life Expectancy (years)")
sns.despine(ax=ax)
```

```python
g = sns.BrokenAxes(ylims=[(0, 7500), (9000, 12500)], height_ratios=[4, 1], hspace=0.02)
g.plot(sns.lineplot, data=healthexp, x="Year", y="Spending_USD", hue="Country")
g.set_axis_labels("Year", "Health Spending (USD)")
```
